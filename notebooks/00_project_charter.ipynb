# The Cost of Discretion: project charter

**Decision problem.** Determine whether formally adjudicated Formula One driving incidents receive comparable treatment after accounting for observable context, and prioritize unusual decisions for human review.

**Primary population.** Race and Sprint adjudications from 2018–2025 involving collisions, forcing another driver off track, gaining an advantage off track, unsafe rejoins, moving under braking, or multiple defensive moves.

**Claims deliberately excluded.** This project cannot detect comparable incidents that Race Control never referred, use a statistical model to determine fault, or prove intentional nationality bias.
